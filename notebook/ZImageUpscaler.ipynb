```python
#@title 🔧 Step 1 — Initialize { display-mode: "form" }

#@markdown ---
#@markdown ### 🔐 GitHub Access Token
#@markdown Required for private repo. Generate at [github.com/settings/tokens](https://github.com/settings/tokens) → `repo` scope
GITHUB_TOKEN = "" #@param {type:"string"}

#@markdown ---

import os, sys, subprocess, warnings
from IPython.display import display, HTML

# ── Theme-safe UI tokens ──
BG_CARD  = '#f4f5f7'
BG_INNER = '#ffffff'
BORDER   = '#dde1e8'
TEXT     = '#2c2c3a'
TEXT_DIM = '#6b7280'
TEXT_MUTED = '#9ca3af'
FONT = "font-family:system-ui,-apple-system,'Segoe UI',Roboto,sans-serif"
MONO = "font-family:'SF Mono',SFMono-Regular,Consolas,monospace"
C_OK   = ('#16a34a', '#f0fdf4', '#15803d')
C_WARN = ('#d97706', '#fffbeb', '#92400e')
C_ERR  = ('#dc2626', '#fef2f2', '#991b1b')
C_INFO = ('#2563eb', '#eff6ff', '#1e40af')
C_STEP = ('#7c3aed', '#f5f3ff', '#5b21b6')

def _card(inner):
    return f'<div style="background:{BG_CARD};border:1px solid {BORDER};border-radius:16px;padding:2px;margin:14px 0"><div style="background:{BG_INNER};border-radius:14px;padding:24px">{inner}</div></div>'

def _header(icon, title, sub=""):
    sub_h = f'<div style="{FONT};font-size:12.5px;color:{TEXT_DIM};margin-top:2px">{sub}</div>' if sub else ''
    return f'<div style="display:flex;align-items:center;gap:12px;margin-bottom:20px;padding-bottom:14px;border-bottom:2px solid {BORDER}"><div style="width:44px;height:44px;display:flex;align-items:center;justify-content:center;background:linear-gradient(135deg,#667eea,#764ba2);border-radius:12px;font-size:22px">{icon}</div><div><div style="{FONT};font-size:20px;font-weight:700;color:{TEXT}">{title}</div>{sub_h}</div></div>'

def _row(msg, colors, icon=""):
    b, bg, t = colors
    return f'<div style="display:flex;align-items:center;gap:10px;padding:10px 14px;margin:6px 0;border-radius:10px;border-left:3px solid {b};background:{bg};color:{t};{MONO};font-size:13.5px">{icon} {msg}</div>'

def ce_show(icon, title, sub=""):
    display(HTML(_card(_header(icon, title, sub))))
def ce_ok(msg, icon="✅"):
    display(HTML(_row(msg, C_OK, icon)))
def ce_warn(msg, icon="⚠️"):
    display(HTML(_row(msg, C_WARN, icon)))
def ce_err(msg, icon="❌"):
    display(HTML(_row(msg, C_ERR, icon)))
def ce_info(msg, icon="ℹ️"):
    display(HTML(_row(msg, C_INFO, icon)))
def ce_step(msg, icon="🔹"):
    display(HTML(_row(msg, C_STEP, icon)))

# ── Suppress warnings ──
warnings.filterwarnings("ignore", category=UserWarning)
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

# ══════════════════════════════════════════════════════
#  SETUP
# ══════════════════════════════════════════════════════

ce_show("🔧", "Initialize", "Installing Real-ESRGAN and downloading model")

# ── GPU check ──
try:
    import subprocess as sp
    has_gpu = sp.run(["nvidia-smi"], capture_output=True, text=True)
    if has_gpu.returncode != 0:
        ce_warn("No GPU detected — Runtime → Change runtime type → T4 GPU")
    else:
        ce_ok("GPU detected")
except Exception:
    ce_warn("Could not check GPU status")

# ── Install deps ──
ce_step("Installing dependencies...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q",
    "realesrgan", "basicsr", "gfpgan", "facexlib", "torch", "torchvision"],
    capture_output=True)
ce_ok("Dependencies installed")

# ── Clone repo (if token provided) ──
if GITHUB_TOKEN.strip():
    ce_step("Cloning repository...")
    repo_url = f"https://{GITHUB_TOKEN.strip()}@github.com/Shineii86/ZImageUpscaler.git"
    result = sp.run(["git", "clone", "-q", repo_url], capture_output=True, text=True)
    if result.returncode == 0:
        os.chdir("ZImageUpscaler")
        ce_ok(f"Repo cloned → {os.getcwd()}")
    else:
        ce_warn(f"Clone failed: {result.stderr.strip()[:100]} — using inline mode")

# ── Download model via aria2c ──
ce_step("Downloading Real-ESRGAN x4plus model (~65 MB)...")
WEIGHTS_DIR = "/content/ZImageUpscaler/weights"
os.makedirs(WEIGHTS_DIR, exist_ok=True)
MODEL_PATH = os.path.join(WEIGHTS_DIR, "RealESRGAN_x4plus.pth")

if not os.path.isfile(MODEL_PATH):
    import subprocess as sp2
    sp2.run(["apt-get", "-y", "install", "-qq", "aria2"], capture_output=True)
    sp2.run([
        "aria2c", "--console-log-level=error", "-c", "-x", "16", "-s", "16", "-k", "1M",
        "-o", "RealESRGAN_x4plus.pth",
        "https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth",
        "-d", WEIGHTS_DIR,
    ], check=True)
    ce_ok(f"Model downloaded ({os.path.getsize(MODEL_PATH) / 1024 / 1024:.0f} MB)")
else:
    ce_ok(f"Model cached ({os.path.getsize(MODEL_PATH) / 1024 / 1024:.0f} MB)")

# ── Download anime model ──
ANIME_MODEL_PATH = os.path.join(WEIGHTS_DIR, "RealESRGAN_x4plus_anime_6B.pth")
if not os.path.isfile(ANIME_MODEL_PATH):
    ce_step("Downloading anime model (~18 MB)...")
    sp2.run([
        "aria2c", "--console-log-level=error", "-c", "-x", "16", "-s", "16", "-k", "1M",
        "-o", "RealESRGAN_x4plus_anime_6B.pth",
        "https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.2.4/RealESRGAN_x4plus_anime_6B.pth",
        "-d", WEIGHTS_DIR,
    ], check=True)
    ce_ok(f"Anime model downloaded")
else:
    ce_ok(f"Anime model cached")

ce_ok("Setup complete!", "🎉")
ce_info("Run the next cell to upscale images", "➡️")
```

```python
#@title 🔍 Step 2 — Upscale Image { display-mode: "form" }

#@markdown ### Settings
#@markdown **Scale** — 2x for moderate, 4x for maximum quality  
#@markdown **Model** — General for photos, Anime for illustrations  
#@markdown **Face Enhance** — GFPGAN for portraits (slower)
scale = "4x" #@param ["2x", "4x"]
model_type = "general" #@param ["general", "anime"]
output_format = "png" #@param ["png", "jpg", "webp"]
face_enhance = False #@param {type:"boolean"}

#@markdown ---

import os, sys, time, gc, uuid
import torch
import numpy as np
from PIL import Image
from IPython.display import display, HTML

# ── Colab file upload ──
try:
    from google.colab import files as colab_files
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# ── UI helpers (re-declare if needed) ──
try:
    ce_show
except NameError:
    BG_CARD='#f4f5f7'; BG_INNER='#ffffff'; BORDER='#dde1e8'; TEXT='#2c2c3a'
    TEXT_DIM='#6b7280'; TEXT_MUTED='#9ca3af'
    FONT="font-family:system-ui,-apple-system,'Segoe UI',Roboto,sans-serif"
    MONO="font-family:'SF Mono',SFMono-Regular,Consolas,monospace"
    C_OK=('#16a34a','#f0fdf4','#15803d'); C_WARN=('#d97706','#fffbeb','#92400e')
    C_ERR=('#dc2626','#fef2f2','#991b1b'); C_INFO=('#2563eb','#eff6ff','#1e40af')
    C_STEP=('#7c3aed','#f5f3ff','#5b21b6')
    def _card(i): return f'<div style="background:{BG_CARD};border:1px solid {BORDER};border-radius:16px;padding:2px;margin:14px 0"><div style="background:{BG_INNER};border-radius:14px;padding:24px">{i}</div></div>'
    def _header(ic,t,s=""):
        sh=f'<div style="{FONT};font-size:12.5px;color:{TEXT_DIM};margin-top:2px">{s}</div>' if s else ''
        return f'<div style="display:flex;align-items:center;gap:12px;margin-bottom:20px;padding-bottom:14px;border-bottom:2px solid {BORDER}"><div style="width:44px;height:44px;display:flex;align-items:center;justify-content:center;background:linear-gradient(135deg,#667eea,#764ba2);border-radius:12px;font-size:22px">{ic}</div><div><div style="{FONT};font-size:20px;font-weight:700;color:{TEXT}">{t}</div>{sh}</div></div>'
    def _row(m,c,ic=""):
        b,gb,tc=c; return f'<div style="display:flex;align-items:center;gap:10px;padding:10px 14px;margin:6px 0;border-radius:10px;border-left:3px solid {b};background:{gb};color:{tc};{MONO};font-size:13.5px">{ic} {m}</div>'
    def ce_show(i,t,s=""): display(HTML(_card(_header(i,t,s))))
    def ce_ok(m,i="✅"): display(HTML(_row(m,C_OK,i)))
    def ce_warn(m,i="⚠️"): display(HTML(_row(m,C_WARN,i)))
    def ce_err(m,i="❌"): display(HTML(_row(m,C_ERR,i)))
    def ce_info(m,i="ℹ️"): display(HTML(_row(m,C_INFO,i)))
    def ce_step(m,i="🔹"): display(HTML(_row(m,C_STEP,i)))

RESULTS_DIR = "/content/ZImageUpscaler/results"
WEIGHTS_DIR = "/content/ZImageUpscaler/weights"
os.makedirs(RESULTS_DIR, exist_ok=True)

# ══════════════════════════════════════════════════════
#  LOAD MODEL
# ══════════════════════════════════════════════════════

ce_show("🔍", "Upscale Image", f"Model: {model_type} · Scale: {scale} · Output: {output_format}")

from basicsr.archs.rrdbnet_arch import RRDBNet
from realesrgan import RealESRGANer

scale_int = int(scale.replace("x", ""))

if model_type == "anime":
    model_file = os.path.join(WEIGHTS_DIR, "RealESRGAN_x4plus_anime_6B.pth")
    net = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=6, num_grow_ch=32, scale=4)
else:
    model_file = os.path.join(WEIGHTS_DIR, "RealESRGAN_x4plus.pth")
    net = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=23, num_grow_ch=32, scale=4)

if not os.path.isfile(model_file):
    ce_err(f"Model not found: {model_file} — re-run Step 1")
    sys.exit(1)

half = torch.cuda.get_device_capability(0)[0] >= 7
upsampler = RealESRGANer(scale=4, model_path=model_file, model=net,
                         tile=0, tile_pad=10, pre_pad=0, half=half, gpu_id=0)
ce_ok("Engine loaded")

# ── Face enhancer ──
face_enhancer = None
if face_enhance:
    ce_step("Loading GFPGAN face enhancer...")
    try:
        from gfpgan import GFPGANer
        face_enhancer = GFPGANer(
            model_path="https://github.com/TencentARC/GFPGAN/releases/download/v1.3.0/GFPGANv1.3.pth",
            upscale=4, arch="clean", channel_multiplier=2, bg_upsampler=None)
        ce_ok("Face enhancer loaded")
    except Exception as e:
        ce_warn(f"GFPGAN failed: {e} — continuing without face enhance")

# ══════════════════════════════════════════════════════
#  UPLOAD & UPSCALE
# ══════════════════════════════════════════════════════

if IN_COLAB:
    display(HTML(f'<div style="border:2px dashed {BORDER};border-radius:14px;padding:28px;text-align:center;background:{BG_CARD};margin:12px 0"><div style="font-size:36px;margin-bottom:8px">📤</div><div style="{FONT};font-size:14px;color:{TEXT};font-weight:500">Upload image(s) to upscale</div><div style="{FONT};font-size:12px;color:{TEXT_MUTED};margin-top:4px">JPG, PNG, WEBP, BMP supported</div></div>'))
    uploaded = colab_files.upload()
    if not uploaded:
        ce_err("No file uploaded")
        sys.exit(0)
    input_files = list(uploaded.keys())
else:
    path = input("Path to image (or directory): ").strip()
    if os.path.isdir(path):
        input_files = [os.path.join(path, f) for f in os.listdir(path)
                       if f.lower().endswith(('.jpg','.jpeg','.png','.webp','.bmp','.tiff'))]
    else:
        input_files = [path]

ce_info(f"{len(input_files)} image(s) to upscale")

# ── Upscale ──
t0 = time.time()
results = []

for i, fpath in enumerate(input_files, 1):
    fn = os.path.basename(fpath)
    ce_step(f"[{i}/{len(input_files)}] {fn}")

    img = Image.open(fpath).convert("RGB")
    img_np = np.array(img)

    gc.collect()
    torch.cuda.empty_cache()

    try:
        with torch.inference_mode():
            output, _ = upsampler.enhance(img_np, outscale=scale_int)
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        ce_warn("OOM — retrying with tile mode")
        upsampler.tile_size = 256
        with torch.inference_mode():
            output, _ = upsampler.enhance(img_np, outscale=scale_int)
        upsampler.tile_size = 0

    # Face enhance
    if face_enhancer is not None:
        try:
            _, _, output = face_enhancer.enhance(img_np, has_aligned=False,
                only_center_face=False, paste_back=True, weight=0.5)
        except Exception as e:
            ce_warn(f"Face enhance failed: {e}")

    result_img = Image.fromarray(output)
    base = os.path.splitext(fn)[0]
    out_fn = f"{base}_{scale}_{uuid.uuid4().hex[:4]}.{output_format}"
    save_path = os.path.join(RESULTS_DIR, out_fn)

    save_kwargs = {}
    if output_format == "jpg": save_kwargs["quality"] = 95
    elif output_format == "webp": save_kwargs["quality"] = 95
    result_img.save(save_path, **save_kwargs)

    w, h = result_img.size
    size_mb = os.path.getsize(save_path) / (1024 ** 2)
    ce_ok(f"{out_fn} — {w}×{h} · {size_mb:.1f} MB")
    results.append((result_img, save_path))

    # Show preview
    display(result_img)

elapsed = time.time() - t0

# ── Summary ──
mins, secs = divmod(int(elapsed), 60)
time_str = f"{mins}m {secs}s" if mins else f"{secs}s"

display(HTML(f'''
<div style="background:linear-gradient(135deg,#667eea 0%,#764ba2 100%);border-radius:14px;padding:20px 24px;color:#ffffff;{FONT};margin:16px 0;box-shadow:0 6px 24px rgba(102,126,234,0.25)">
  <div style="font-size:18px;font-weight:700;margin-bottom:4px">🎉 Upscale complete!</div>
  <div style="font-size:13px;opacity:0.9">{len(results)} image(s) upscaled in {time_str}</div>
</div>
'''))

stats_html = (f'<div style="background:{BG_INNER};border:1px solid {BORDER};border-radius:12px;padding:16px;text-align:center;min-width:120px;flex:1"><div style="{FONT};font-size:26px;font-weight:800;color:#16a34a">{len(results)}</div><div style="{FONT};font-size:11px;color:{TEXT_MUTED};text-transform:uppercase;letter-spacing:0.8px;font-weight:600;margin-top:4px">Upscaled</div></div>'
    + f'<div style="background:{BG_INNER};border:1px solid {BORDER};border-radius:12px;padding:16px;text-align:center;min-width:120px;flex:1"><div style="{FONT};font-size:26px;font-weight:800;color:#2563eb">{scale}</div><div style="{FONT};font-size:11px;color:{TEXT_MUTED};text-transform:uppercase;letter-spacing:0.8px;font-weight:600;margin-top:4px">Scale</div></div>'
    + f'<div style="background:{BG_INNER};border:1px solid {BORDER};border-radius:12px;padding:16px;text-align:center;min-width:120px;flex:1"><div style="{FONT};font-size:26px;font-weight:800;color:#2563eb">{time_str}</div><div style="{FONT};font-size:11px;color:{TEXT_MUTED};text-transform:uppercase;letter-spacing:0.8px;font-weight:600;margin-top:4px">Elapsed</div></div>')
display(HTML(f'<div style="display:flex;gap:12px;margin:14px 0;flex-wrap:wrap">{stats_html}</div>'))

# Cleanup uploaded files
if IN_COLAB:
    for f in input_files:
        if os.path.exists(f):
            os.remove(f)
```
